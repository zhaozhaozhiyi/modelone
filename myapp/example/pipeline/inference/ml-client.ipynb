```python
# 弹性伸缩测试

import requests
from concurrent.futures import ThreadPoolExecutor

# 定义要启动的线程数（即客户端数）
num_clients = 1
request_num = 100
try_num=0
def load_test():
    global try_num
    # 发送GET请求
    url = "http://decisiontree-202310011.service/v1/models/decisiontree/metadata"
    headers = {
      'Authorization': 'admin',
      'Content-Type': 'application/json'
    }
    response = requests.request("GET", url, headers=headers)
    # 打印响应状态码
    print(try_num,'成功' if response.status_code==200 else '失败',json.loads(response.text).get("model_spec",{}).get("version",'未知'),flush=True)
    
    try_num+=1
load_test()
# 创建一个ThreadPoolExecutor实例
with ThreadPoolExecutor(max_workers=num_clients) as executor:
    # 使用executor提交load_test任务，参数为url
    for _ in range(request_num):
        executor.submit(load_test)
```
